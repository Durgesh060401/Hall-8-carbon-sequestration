```python
# ============================================================
# HALL-8 CARBON SEQUESTRATION ANALYSIS
# IIT KANPUR
#
# CSV VERSION
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.4f}".format)


# ============================================================
# 2. CSV FILE PATH
# ============================================================

from pathlib import Path

DATA_FILE = "Primary_data_Hall_8_iit_Kanpur - Copy of Tree data.csv"

# ------------------------------------------------------------
# Find the data file automatically
# ------------------------------------------------------------

# GitHub repository structure:
# Hall-8-carbon-sequestration/
# ├── Code/
# │   └── Hall8_Carbon_Sequestration_Analysis.ipynb
# └── Data/
#     └── Primary_data_Hall_8_iit_Kanpur - Copy of Tree data.csv

github_path = Path("../Data") / DATA_FILE

# Google Colab path
colab_path = Path("/content") / DATA_FILE


if github_path.exists():

    DATA_PATH = github_path

elif colab_path.exists():

    DATA_PATH = colab_path

else:

    raise FileNotFoundError(
        f"Could not find '{DATA_FILE}'. "
        "Please check that the CSV exists in the Data folder."
    )


print("Using data file:")
print(DATA_PATH)

# ============================================================
# 3. LOAD CSV
# ============================================================

df = pd.read_csv(DATA_PATH)

print("CSV loaded successfully.")

print("\nDataset shape:")
print(df.shape)


# ============================================================
# 4. DISPLAY THE EXACT ORIGINAL COLUMN NAMES
# ============================================================

print("\n========================================")
print("ORIGINAL CSV COLUMN NAMES")
print("========================================")

for i, col in enumerate(df.columns, start=1):
    print(i, repr(col))


# ============================================================
# 5. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
)


# ============================================================
# 6. FUNCTION TO FIND COLUMNS
# ============================================================

def find_column(columns, keywords, exclude=None):

    """
    Find a column based on keywords.

    Example:
        find_column(df.columns, ["adjusted", "dbh"])

    searches for a column containing both words.
    """

    if exclude is None:
        exclude = []

    for col in columns:

        col_lower = (
            str(col)
            .strip()
            .lower()
        )

        # Skip excluded columns
        if any(
            ex.lower() in col_lower
            for ex in exclude
        ):
            continue

        if all(
            keyword.lower() in col_lower
            for keyword in keywords
        ):
            return col

    return None


# ============================================================
# 7. AUTOMATICALLY FIND IMPORTANT COLUMNS
# ============================================================

columns = df.columns


# ------------------------------------------------------------
# Block / pathway
# ------------------------------------------------------------

block_col = find_column(
    columns,
    ["block"]
)

if block_col is None:
    block_col = find_column(
        columns,
        ["pathway"]
    )


# ------------------------------------------------------------
# Common name
# ------------------------------------------------------------

common_col = find_column(
    columns,
    ["general", "name"]
)

if common_col is None:
    common_col = find_column(
        columns,
        ["common", "name"]
    )


# ------------------------------------------------------------
# Scientific name
# ------------------------------------------------------------

scientific_col = find_column(
    columns,
    ["scientific", "name"]
)


# ------------------------------------------------------------
# Trunks
# ------------------------------------------------------------

trunks_col = find_column(
    columns,
    ["trunk"]
)


# ------------------------------------------------------------
# Circumference
# ------------------------------------------------------------

circumference_col = find_column(
    columns,
    ["circumference"]
)


# ------------------------------------------------------------
# DBH
# ------------------------------------------------------------

dbh_col = find_column(
    columns,
    ["dbh"],
    exclude=["adjusted"]
)


# ------------------------------------------------------------
# ADJUSTED DBH
# ------------------------------------------------------------

adjusted_dbh_col = find_column(
    columns,
    ["adjusted", "dbh"]
)


# ------------------------------------------------------------
# HEIGHT
# ------------------------------------------------------------

height_col = find_column(
    columns,
    ["height"]
)

# If the spreadsheet contains the typo "Hight"
if height_col is None:

    height_col = find_column(
        columns,
        ["hight"]
    )


# ------------------------------------------------------------
# DENSITY
# ------------------------------------------------------------

density_col = find_column(
    columns,
    ["density"]
)


# ------------------------------------------------------------
# AGB STEM
# ------------------------------------------------------------

agb_stem_col = find_column(
    columns,
    ["agb", "stem"]
)


# ------------------------------------------------------------
# AGB FOLIAGE
# ------------------------------------------------------------

agb_foliage_col = find_column(
    columns,
    ["agb", "foliage"]
)


# ------------------------------------------------------------
# BGB
# ------------------------------------------------------------

bgb_col = find_column(
    columns,
    ["bgb"]
)


# ------------------------------------------------------------
# TOTAL BIOMASS
# ------------------------------------------------------------

tb_col = find_column(
    columns,
    ["tb"]
)

if tb_col is None:

    tb_col = find_column(
        columns,
        ["total", "biomass"]
    )


# ============================================================
# 8. SHOW WHAT WAS FOUND
# ============================================================

print("\n========================================")
print("COLUMN MATCHING")
print("========================================")

print("Block             :", block_col)
print("Common name       :", common_col)
print("Scientific name   :", scientific_col)
print("Trunks            :", trunks_col)
print("Circumference     :", circumference_col)
print("DBH               :", dbh_col)
print("Adjusted DBH      :", adjusted_dbh_col)
print("Height            :", height_col)
print("Density           :", density_col)
print("AGB Stem          :", agb_stem_col)
print("AGB Foliage       :", agb_foliage_col)
print("BGB               :", bgb_col)
print("Total Biomass     :", tb_col)


# ============================================================
# 9. CHECK REQUIRED COLUMNS
# ============================================================

required_matches = {

    "block_pathway": block_col,
    "common_name": common_col,
    "scientific_name": scientific_col,
    "trunks": trunks_col,
    "circumference_cm": circumference_col,
    "dbh_cm": dbh_col,
    "adjusted_dbh_cm": adjusted_dbh_col,
    "height_m": height_col,
    "density": density_col,
    "agb_stem_kg": agb_stem_col,
    "agb_foliage_kg": agb_foliage_col,
    "bgb_kg": bgb_col,
    "total_biomass_kg": tb_col
}


missing = [

    name
    for name, original
    in required_matches.items()
    if original is None

]


if missing:

    print("\n========================================")
    print("COLUMN MATCHING FAILED")
    print("========================================")

    print("\nThe following variables could not be identified:")

    for item in missing:
        print("-", item)

    print("\nActual CSV columns are:")

    for col in df.columns:
        print("-", repr(col))

    raise KeyError(
        f"Could not identify columns: {missing}"
    )


# ============================================================
# 10. RENAME EVERYTHING TO STANDARD NAMES
# ============================================================

rename_dictionary = {

    block_col: "block_pathway",

    common_col: "common_name",

    scientific_col: "scientific_name",

    trunks_col: "trunks",

    circumference_col: "circumference_cm",

    dbh_col: "dbh_cm",

    adjusted_dbh_col: "adjusted_dbh_cm",

    height_col: "height_m",

    density_col: "density",

    agb_stem_col: "agb_stem_kg",

    agb_foliage_col: "agb_foliage_kg",

    bgb_col: "bgb_kg",

    tb_col: "total_biomass_kg"

}


df = df.rename(
    columns=rename_dictionary
)


print("\n========================================")
print("STANDARDIZATION SUCCESSFUL")
print("========================================")

print(df.columns.tolist())


# ============================================================
# 11. REMOVE EMPTY ROWS
# ============================================================

df = df.dropna(
    how="all"
).copy()


# ============================================================
# 12. CLEAN TEXT VARIABLES
# ============================================================

text_columns = [

    "block_pathway",
    "common_name",
    "scientific_name"

]


for col in text_columns:

    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
    )


# ============================================================
# 13. CONVERT NUMERICAL VARIABLES
# ============================================================

numeric_columns = [

    "trunks",

    "circumference_cm",

    "dbh_cm",

    "adjusted_dbh_cm",

    "height_m",

    "density",

    "agb_stem_kg",

    "agb_foliage_kg",

    "bgb_kg",

    "total_biomass_kg"

]


for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ============================================================
# 14. BASIC DATA CHECK
# ============================================================

print("\n========================================")
print("DATA CHECK")
print("========================================")

print(
    "Number of trees:",
    len(df)
)

print(
    "Number of columns:",
    len(df.columns)
)

print("\nMissing values:")

display(
    df.isna().sum()
)


# ============================================================
# 15. NUMBER OF SPECIES
# ============================================================

total_trees = len(df)

number_of_species = (
    df["scientific_name"]
    .nunique()
)


print("\n========================================")
print("TREE INVENTORY")
print("========================================")

print(
    "Number of trees:",
    total_trees
)

print(
    "Number of species:",
    number_of_species
)


# ============================================================
# 16. CALCULATE DBH FROM CIRCUMFERENCE
# ============================================================

# DBH = Circumference / π

df["dbh_calculated_cm"] = (

    df["circumference_cm"]
    /
    np.pi

)


# ============================================================
# 17. DBH VALIDATION
# ============================================================

df["dbh_difference"] = (

    df["dbh_cm"]
    -
    df["dbh_calculated_cm"]

)


print("\n========================================")
print("DBH VALIDATION")
print("========================================")

print(
    "Maximum absolute DBH difference:",
    df["dbh_difference"]
    .abs()
    .max()
)


display(

    df[
        [
            "circumference_cm",
            "dbh_cm",
            "dbh_calculated_cm",
            "dbh_difference"
        ]
    ].head(10)

)


# ============================================================
# 18. USE ADJUSTED DBH
# ============================================================

# IMPORTANT:
#
# Your Google Sheet uses Adjusted DBH for the AGB calculation.
#
# Therefore:
#
# adjusted DBH (cm)
#          ↓
# adjusted DBH (m)
#          ↓
# AGB


df["dbh_used_cm"] = (
    df["adjusted_dbh_cm"]
)


df["dbh_used_m"] = (

    df["dbh_used_cm"]
    /
    100

)


# ============================================================
# 19. CHAVE AGB STEM
# ============================================================

# Formula:
#
# AGB =
#
# 0.0673 ×
# (Density × D² × Height)^0.976
#
#
# D = adjusted DBH in metres


df["agb_stem_calculated_kg"] = (

    0.0673
    *
    (
        df["density"]
        *
        (
            df["dbh_used_m"]
            ** 2
        )
        *
        df["height_m"]
    )
    ** 0.976

)


# ============================================================
# 20. AGB VALIDATION
# ============================================================

df["agb_stem_difference"] = (

    df["agb_stem_kg"]
    -
    df["agb_stem_calculated_kg"]

)


print("\n========================================")
print("AGB STEM VALIDATION")
print("========================================")

print(
    "Maximum absolute AGB difference:",
    df["agb_stem_difference"]
    .abs()
    .max()
)


display(

    df[
        [
            "scientific_name",

            "circumference_cm",

            "dbh_cm",

            "adjusted_dbh_cm",

            "height_m",

            "density",

            "agb_stem_kg",

            "agb_stem_calculated_kg",

            "agb_stem_difference"

        ]
    ].head(15)

)


# ============================================================
# 21. FOLIAGE BIOMASS
# ============================================================

# Foliage biomass = 4% of stem AGB

FOLIAGE_FRACTION = 0.04


df["agb_foliage_calculated_kg"] = (

    FOLIAGE_FRACTION
    *
    df["agb_stem_calculated_kg"]

)


# ============================================================
# 22. TOTAL ABOVE-GROUND BIOMASS
# ============================================================

df["agb_total_kg"] = (

    df["agb_stem_calculated_kg"]
    +
    df["agb_foliage_calculated_kg"]

)


# ============================================================
# 23. BELOW-GROUND BIOMASS
# ============================================================

# BGB = 20% of AGB

BGB_FRACTION = 0.20


df["bgb_calculated_kg"] = (

    BGB_FRACTION
    *
    df["agb_total_kg"]

)


# ============================================================
# 24. TOTAL BIOMASS
# ============================================================

df["total_biomass_calculated_kg"] = (

    df["agb_total_kg"]
    +
    df["bgb_calculated_kg"]

)


# ============================================================
# 25. CARBON STOCK
# ============================================================

# Carbon fraction = 0.47

CARBON_FRACTION = 0.47


df["carbon_stock_kg"] = (

    CARBON_FRACTION
    *
    df["total_biomass_calculated_kg"]

)


# ============================================================
# 26. CO2 EQUIVALENT
# ============================================================

# CO2 = Carbon × 44/12

CO2_CONVERSION = 44 / 12


df["co2e_kg"] = (

    df["carbon_stock_kg"]
    *
    CO2_CONVERSION

)


# ============================================================
# 27. FINAL TREE-LEVEL DATA
# ============================================================

analysis_df = df[

    [
        "block_pathway",

        "common_name",

        "scientific_name",

        "trunks",

        "circumference_cm",

        "dbh_cm",

        "adjusted_dbh_cm",

        "height_m",

        "density",

        "agb_stem_calculated_kg",

        "agb_foliage_calculated_kg",

        "agb_total_kg",

        "bgb_calculated_kg",

        "total_biomass_calculated_kg",

        "carbon_stock_kg",

        "co2e_kg"

    ]

].copy()


# ============================================================
# 28. OVERALL RESULTS
# ============================================================

total_agb = (
    df["agb_total_kg"]
    .sum()
)

total_bgb = (
    df["bgb_calculated_kg"]
    .sum()
)

total_biomass = (
    df["total_biomass_calculated_kg"]
    .sum()
)

total_carbon = (
    df["carbon_stock_kg"]
    .sum()
)

total_co2e = (
    df["co2e_kg"]
    .sum()
)


print("\n============================================")
print("OVERALL CARBON STOCK RESULTS")
print("============================================")

print(
    f"Number of trees          : {total_trees}"
)

print(
    f"Number of species        : {number_of_species}"
)

print(
    f"Above-ground biomass     : {total_agb:.3f} kg"
)

print(
    f"Below-ground biomass     : {total_bgb:.3f} kg"
)

print(
    f"Total biomass            : {total_biomass:.3f} kg"
)

print(
    f"Carbon stock             : {total_carbon:.3f} kg C"
)

print(
    f"CO2 equivalent           : {total_co2e:.3f} kg CO2e"
)


# ============================================================
# 29. SPECIES-LEVEL ANALYSIS
# ============================================================

species_summary = (

    df

    .groupby(
        [
            "common_name",
            "scientific_name"
        ],
        dropna=False
    )

    .agg(

        trees=(
            "scientific_name",
            "size"
        ),

        total_agb_kg=(
            "agb_total_kg",
            "sum"
        ),

        total_bgb_kg=(
            "bgb_calculated_kg",
            "sum"
        ),

        total_biomass_kg=(
            "total_biomass_calculated_kg",
            "sum"
        ),

        carbon_stock_kg=(
            "carbon_stock_kg",
            "sum"
        ),

        co2e_kg=(
            "co2e_kg",
            "sum"
        )

    )

    .reset_index()

)


# ============================================================
# 30. CARBON SHARE
# ============================================================

species_summary["carbon_share_percent"] = (

    species_summary["carbon_stock_kg"]
    /
    total_carbon
    *
    100

)


# ============================================================
# 31. SORT SPECIES
# ============================================================

species_summary = (

    species_summary

    .sort_values(
        "carbon_stock_kg",
        ascending=False
    )

    .reset_index(
        drop=True
    )

)


print("\n========================================")
print("SPECIES-LEVEL RESULTS")
print("========================================")

display(
    species_summary
)


# ============================================================
# 32. PROJECT SUMMARY
# ============================================================

project_summary = pd.DataFrame({

    "Indicator": [

        "Number of trees",

        "Number of species",

        "Above-ground biomass",

        "Below-ground biomass",

        "Total biomass",

        "Carbon stock",

        "CO2 equivalent"

    ],

    "Value": [

        total_trees,

        number_of_species,

        total_agb,

        total_bgb,

        total_biomass,

        total_carbon,

        total_co2e

    ],

    "Unit": [

        "trees",

        "species",

        "kg",

        "kg",

        "kg",

        "kg C",

        "kg CO2e"

    ]

})


print("\n========================================")
print("PROJECT SUMMARY")
print("========================================")

display(
    project_summary
)


# ============================================================
# 33. TREE DISTRIBUTION BY SPECIES
# ============================================================

species_count = (

    df["common_name"]

    .value_counts()

    .sort_values(
        ascending=True
    )

)


plt.figure(
    figsize=(10, 7)
)

plt.barh(
    species_count.index,
    species_count.values
)

plt.xlabel(
    "Number of Trees"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Tree Inventory by Species"
)

plt.tight_layout()

plt.show()


# ============================================================
# 34. CARBON STOCK BY SPECIES
# ============================================================

carbon_plot = (

    species_summary

    .sort_values(
        "carbon_stock_kg",
        ascending=True
    )

)


plt.figure(
    figsize=(10, 7)
)

plt.barh(

    carbon_plot["common_name"],

    carbon_plot["carbon_stock_kg"]

)

plt.xlabel(
    "Carbon Stock (kg C)"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Estimated Carbon Stock by Species"
)

plt.tight_layout()

plt.show()


# ============================================================
# 35. CO2 EQUIVALENT BY SPECIES
# ============================================================

co2_plot = (

    species_summary

    .sort_values(
        "co2e_kg",
        ascending=True
    )

)


plt.figure(
    figsize=(10, 7)
)

plt.barh(

    co2_plot["common_name"],

    co2_plot["co2e_kg"]

)

plt.xlabel(
    "Stored CO2 Equivalent (kg CO2e)"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Stored CO2 Equivalent by Species"
)

plt.tight_layout()

plt.show()


# ============================================================
# 36. TOTAL BIOMASS BY SPECIES
# ============================================================

biomass_plot = (

    species_summary

    .sort_values(
        "total_biomass_kg",
        ascending=True
    )

)


plt.figure(
    figsize=(10, 7)
)

plt.barh(

    biomass_plot["common_name"],

    biomass_plot["total_biomass_kg"]

)

plt.xlabel(
    "Total Biomass (kg)"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Estimated Total Biomass by Species"
)

plt.tight_layout()

plt.show()


# ============================================================
# 37. CREATE RESULTS FOLDER
# ============================================================

results_dir = Path(
    "/content/results"
)

results_dir.mkdir(
    exist_ok=True
)


figures_dir = (
    results_dir
    /
    "figures"
)

figures_dir.mkdir(
    exist_ok=True
)


# ============================================================
# 38. SAVE TREE-LEVEL RESULTS
# ============================================================

analysis_df.to_csv(

    results_dir
    /
    "tree_level_carbon_analysis.csv",

    index=False

)


# ============================================================
# 39. SAVE SPECIES RESULTS
# ============================================================

species_summary.to_csv(

    results_dir
    /
    "species_carbon_stock.csv",

    index=False

)


# ============================================================
# 40. SAVE PROJECT SUMMARY
# ============================================================

project_summary.to_csv(

    results_dir
    /
    "project_summary.csv",

    index=False

)


# ============================================================
# 41. GOOGLE SHEET VS PYTHON VALIDATION
# ============================================================

validation = pd.DataFrame({

    "Variable": [

        "AGB Stem",

        "BGB",

        "Total Biomass"

    ],

    "Google_Sheet": [

        df["agb_stem_kg"].sum(),

        df["bgb_kg"].sum(),

        df["total_biomass_kg"].sum()

    ],

    "Python": [

        df["agb_stem_calculated_kg"].sum(),

        df["bgb_calculated_kg"].sum(),

        df["total_biomass_calculated_kg"].sum()

    ]

})


validation["Difference"] = (

    validation["Python"]
    -
    validation["Google_Sheet"]

)


print("\n========================================")
print("GOOGLE SHEET VS PYTHON VALIDATION")
print("========================================")

display(
    validation
)


# ============================================================
# 42. SAVE VALIDATION
# ============================================================

validation.to_csv(

    results_dir
    /
    "google_sheet_python_validation.csv",

    index=False

)


# ============================================================
# 43. SAVE FIGURES
# ============================================================

plt.figure(
    figsize=(10, 7)
)

plt.barh(
    species_count.index,
    species_count.values
)

plt.xlabel(
    "Number of Trees"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Tree Inventory by Species"
)

plt.tight_layout()

plt.savefig(

    figures_dir
    /
    "species_distribution.png",

    dpi=300,

    bbox_inches="tight"

)

plt.close()


# ------------------------------------------------------------
# Carbon
# ------------------------------------------------------------

plt.figure(
    figsize=(10, 7)
)

plt.barh(

    carbon_plot["common_name"],

    carbon_plot["carbon_stock_kg"]

)

plt.xlabel(
    "Carbon Stock (kg C)"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Estimated Carbon Stock by Species"
)

plt.tight_layout()

plt.savefig(

    figures_dir
    /
    "carbon_stock_by_species.png",

    dpi=300,

    bbox_inches="tight"

)

plt.close()


# ------------------------------------------------------------
# CO2
# ------------------------------------------------------------

plt.figure(
    figsize=(10, 7)
)

plt.barh(

    co2_plot["common_name"],

    co2_plot["co2e_kg"]

)

plt.xlabel(
    "Stored CO2 Equivalent (kg CO2e)"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Stored CO2 Equivalent by Species"
)

plt.tight_layout()

plt.savefig(

    figures_dir
    /
    "co2_equivalent_by_species.png",

    dpi=300,

    bbox_inches="tight"

)

plt.close()


# ------------------------------------------------------------
# Biomass
# ------------------------------------------------------------

plt.figure(
    figsize=(10, 7)
)

plt.barh(

    biomass_plot["common_name"],

    biomass_plot["total_biomass_kg"]

)

plt.xlabel(
    "Total Biomass (kg)"
)

plt.ylabel(
    "Species"
)

plt.title(
    "Estimated Total Biomass by Species"
)

plt.tight_layout()

plt.savefig(

    figures_dir
    /
    "biomass_by_species.png",

    dpi=300,

    bbox_inches="tight"

)

plt.close()


# ============================================================
# 44. FINAL OUTPUT
# ============================================================

print("\n")
print("============================================")
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("============================================")

print(
    f"Trees analysed       : {total_trees}"
)

print(
    f"Species identified   : {number_of_species}"
)

print(
    f"Total biomass        : {total_biomass:.3f} kg"
)

print(
    f"Carbon stock         : {total_carbon:.3f} kg C"
)

print(
    f"CO2 equivalent       : {total_co2e:.3f} kg CO2e"
)

print("\nResults saved in:")
print(results_dir)

print("\nOutput files:")
print("1. tree_level_carbon_analysis.csv")
print("2. species_carbon_stock.csv")
print("3. project_summary.csv")
print("4. google_sheet_python_validation.csv")

print("\nFigures:")
print("1. species_distribution.png")
print("2. carbon_stock_by_species.png")
print("3. co2_equivalent_by_species.png")
print("4. biomass_by_species.png")
```
